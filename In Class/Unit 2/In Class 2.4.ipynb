# **Common Dictionary Methods**

*   .pop() - removes the item with the specified key
*   .update() - adds or changes dictionary items
*   .clear() - removes all items from dictionary
*   .get() - returns value of specified key
*   .popitem() - returns the last inserted key and value pair as a tuple and removes it from the dictionary
*   .copy() - returns a copy of the dictionary

```python
# this is an example of a dictionary, notice the key:value pairs
# to access a value in the dictionary, place the key inside square brackets following the name of the dictionary

mydict = {
    'Eggs':2.59,
    'Milk':3.19,
    'Cheese':4.80,
    'Yogurt':[1.35,3.50],
    'Butter':2.59,
}
```

Access the value associated with cheese using the dictionary and print it out.

```python
print(f'The price of cheese is ${mydict['Cheese']}')
```

Do the same thing using get() with the key inside of the parentheses and print it out.

```python
print(f'The price of cheese is ${mydict.get('Cheese')}')
```

Access the second value in the list associated with yogurt and print it out.

```python
print(mydict['Yogurt'][1])
```

Change the value associated with butter to something else and print the dictionary.

```python
mydict.update({'Butter':3.50})
print(mydict)
```

Add a food and its associated price directly to the dictionary and print the dictionary.

```python
mydict.update({'Apples':1.50})
print(mydict)
```

# **Looping Through Dictionaries**

*   .items() - returns a list of key-value pairs
*   .keys() - returns a list of keys
*   .values() - returns a list of values

Because dictionaries can contain large amounts of data, it is sometimes useful to loop through them.

```python
material_costs = {
    "Concrete": 150,    # Cost per cubic yard
    "Steel": 700,       # Cost per ton
    "Brick": 0.50,      # Cost per brick
    "Drywall": 12,      # Cost per sheet
    "Asphalt": 100,     # Cost per ton
    "Glass": 25,        # Cost per square foot
    "Insulation": 0.80, # Cost per square foot
    "Rebar": 500        # Cost per ton
}
```

Add an input statement that asks what the user would like to buy, and let the user know if it is in the dictionary or not (use an if statement)

```python
materials=input("What would you like to buy?")
for material in material_costs:
  if materials in material_costs.keys():
    print("Material is in dictionary")
    break
else:
    print("Material is not in dictionary")

```

Add a while loop to keep asking them for a new material until they enter 'quit' as the material

```python
materials = input("What would you like to buy? ")

while materials != 'quit':
    if materials in material_costs:
        print("Material is in dictionary")
    else:
        print("Material is not in dictionary")

    materials = input("What would you like to buy? ")
```

You can loop through a list of the keys and values by using .items()<br>
Loop through the dictionary above and print out how much each material costs.

```python
for key, value in material_costs.items():
  print(f'The price of {key} is ${value:2}')
```

Repeat the steps above, but this time omit the costs of brick and glass using if statements.

```python
for key, value in material_costs.items():
  if key != 'Brick' and key != 'Glass':
    print(f'The price of {key} is ${value:2}')
```

You can loop through just the keys by using .keys(). Loop through the keys in the dictionary and print the key-value pairs.

```python
for key in material_costs.keys():
  print(f'The price of {key} is ${material_costs[key]:2}')
```

You can loop through just the values by using .values(). Loop through the values in the dictionary and print the value.

```python
print('PRICE DUMP:')
for valuein in material_costs.values():
  print(f'${valuein:2}')
```

You can loop through the keys and access the values by square brackets with the key. Loop through the keys, access the values and print the values using the dict[key] construction

```python
for key in material_costs.keys():
  print(f'The price of {key} is ${material_costs[key]:2}')
```

```python
def convert(inputSentence):
  myList = inputSentence.lower().split(" ")
  return myList

redneckTranslator = {
  'with': 'wif',
  'hear':'hyar',
  'from': 'frum',
  'are': 'is',
  'saw': 'seen',
  'get': 'git',
  'wanting': 'hankerin',
  'think': 'reckon',
  'planning': "fixin'",
  'people': 'folks',
  'call': 'holler',
  'something': 'sumpin',
  'never': "ain't never",
  'naked': 'nekkid',
  'wash': 'warsh',
  'beautiful': 'purty',
  'am not': "ain't",
  'are not': "ain't",
  "isn't": "ain't",
  "aren't": "ain't",
  'is not': "ain't",
  'is no': "ain't no",
  'rude': 'ignert',
  'were': 'was',
  'you': 'ya',
  'theater': 'show',
  'time': 'spell',
  'negotiate': 'dicker',
  'contract': 'deal',
  'woman': 'gal',
  'man': 'feller',
  'teenager': 'varmint',
  'before': "b'fo'",
  'minutes': 'a jiffy',
  'ing': "in'",
  'that': 'that there',
  'for': 'fur',
  'this': 'tis',
  'am': 'is',
  'to': "t'"
}

mySentence = input("Enter your sentence here:")

listSentence = convert(mySentence)

newSentence = ""
for origWord in listSentence:
  if origWord in redneckTranslator.keys():
    newWord = redneckTranslator[origWord]
    newSentence += newWord + " "
  else:
    newSentence += origWord + " "

print(newSentence)
```
