# Nested For Loops

````python
# Run this code to combine adjectives and nouns using a nested loop

adjs = ["large", "happy", "fat", "quiet"]
animals = ["zebra", "porcupine", "elephant", "giraffe"]

for adj in adjs:
  for animal in animals:
    print(adj, animal)
````

````python
# Create your own nested for loop using the lists below

adjs = ["sweet", "ripe", "delicious"]
fruits = ["pear", "mango", "kiwi"]

for adj in adjs:
  for fruit in fruits:
    print(adj,fruit)
````

# Creating a list within a loop

````python
# Run the following code

mylist = ['Amy', 'Anne', 'Barbara', 'Bob', 'Clare', 'Clark', 'Jane', 'John', 'Jenny', 'Jerry', 'Joe']

myNewList = []

for i in mylist:
  newItem = i + ' Smith'
  myNewList.append(newItem)

print(myNewList)
````

````python
# In this code block, write your own for loop using mylist
# Also, create a list that tells everyone Hello "Name" for each name
# Create a list of names
mylist = ['Amy', 'Anne', 'Barbara', 'Bob', 'Clare', 'Clark', 'Jane', 'John', 'Jenny', 'Jerry', 'Joe']

# Create an empty list to store greetings
helloList = []

# Use a for loop to build the new list
for name in mylist:
    greeting = "Hello " + name
    helloList.append(greeting)

# Print each greeting on its own line
for greeting in helloList:
    print(greeting)
````

# Lists of Lists in For loops
When you have a list of lists we learned how to access the various elements.

For example if you have a list of fruits


```
# This is formatted as code
fruits = [["banana", "yellow"], ["apple","red"], ["grape","blue"]]
```
We can access the color of an apple as


```
# This is formatted as code
fruits[1][1]
```
Or the second element of the second list

We can also take elements from the list into a new variable.


```
# This is formatted as code
fruit, color = fruits[1]
```
This takes the second element in the list and sets the variable fruit="apple" and the variable color="red"


We can do something similar in loops, but there is a neat shortcut we will show you.

Try the below activity, then we will show you the trick

````python
# Working with "lists of lists" is similar - you just need to remember each item in the list is a list

names = [
    ["Gus", "Williams"],
    ["Norm", "Jones"],
    ["Brandan", "Williamson"]
    ]

# Note: you could put the above all on one line, but it is more readable to use multiple lines
# python realizes that you haven't completed the code, and allows you to have more than one line

first, last = names[0]
print(first, last)
````

````python
# Print the first list from "names"

print(names[0])
````

````python
# Write a print statement for the first list in names that says:
# my first name is [fill in the name here] and my last name is [fill in the name here]

print(f'My first name is{names[0][0]}and my last name is {names[0][1]}.')
````

````python
# Now write a loop that goes through the list and prints the statement from above for each name

for name in names:
  print(f'My first name is {name[0]} and my last name is {name[1]}.')
````

````python
# There is another way to work with a list of lists in loops using multiple variables
# fname, lname vs. name[0] name[1]

for first, last in names:
  print(f'My first name is {first} and my last name is {last}.')
````

# Slicing

Now we will experiment with using slicing to access a portion of a list.

````python
# Run this cell to create a new list of cities to visit
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Philadelphia', 'San Antonio']
````

````python
# Use slicing to print the first 3 cities in the list (note how the index works)

print(cities[0:3])
````

````python
# Use slicing to print the 3rd and 4th cities in the list

print(cities[2:4])
````

````python
# Use slicing to print the cities from the 4th city to the end of the list.

print(cities[3:])
````

````python
# Use a negative slice index to print the last two cities

print(cities[-2:])
````

````python
# Run this code to create a new list.
cities2 = cities
print(cities)
print(cities2)
````

````python
# Now append 'Provo' to the original list and print both lists. Notice what happens
cities.append('Provo')
print(cities)
print(cities2)

````

````python
# Now write a statement that makes a new list called cities3 which is a copy of the cities list.
# This is separate list created using slicing.
# Modify the new list and print both lists.

cities2=cities[:]

cities2.append('Richfield')
print(cities2)
print(cities)
````

# Range Function
Now we will use the range function with loops to create some lists of numbers

````python
# These are what we call functions. Similar to the range function.
# We will talk about these ones another day, but try to figure out what they do and run this code block.
def sum_avg(numbers):
  total = sum(numbers)
  average = total / len(numbers)
  return total, average
````

````python
# Write a for loop with the range function to create a list of all the numbers between 1 and 5
number1to5 = []

for i in range(1,6):
  print(i)
  number1to5.append(i)

print(number1to5)



print(f'The total of these numbers is {sum_avg(number1to5)[0]} and the average is {sum_avg(number1to5)[1]}')
````

````python
# Write a for loop with the range function to create a list of squares of all the numbers between 1 and 5
number1to5squared = []

for i in range(1,6):
  number1to5squared.append(i ** 2)
  print(i ** 2)


print(f'The total of these numbers is {sum_avg(number1to5squared)[0]} and the average is {sum_avg(number1to5squared)[1]}')
````

````python
# Write a for loop with the range function to create a list of all the multiples of 20 between 1 and 200
multiple20 = []

for i in range (20,201,20):
  multiple20.append(i)
  print(i)



print(f'The total of these numbers is {sum_avg(multiple20)[0]} and the average is {sum_avg(multiple20)[1]}')
````
