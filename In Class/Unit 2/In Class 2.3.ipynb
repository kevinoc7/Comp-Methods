Create a variable to store what your favorite U.S. state is.

```python
favorite_state='kansas'
```

Check if the user's favorite state is in the list of all states, capitalizes all letters in that state, capitalizes the first letter in all other states, then print the modified list. (Hint: Use a for loop to check the list)

```python
modified_states=[]
for state in states:
  if state==favorite_state:
    modified_states.append(state.upper())
  else:
    modified_states.append(state.capitalize())
print(modified_states)
```

In the next code block, use the variables from the form. First, print a greeting to the user using their name, then tell them whether or not they are eligible to take the professional engineering (PE) licensing exam. To be eligible, the following conditions must all apply:

1) Graduate from BYU (not really, but let's roll with it).
2) At least 4 years of work experience
3) Have passed the fundamentals of engineering (FE) exam.

If the user is eligible, print a message and stop. If the user is not eligible, print a message and then list the reasons why the person is not eligible.

```python
# @title PE Eligibility form
my_name = "Kevin Campbell" #@param {type:"string"}
college = "BYU" #@param ["USU", "Utah", "BYU"]
years_exp = 6 #@param {type:"slider", min:0, max:10, step:1}
pass_fe = False #@param {type:"boolean"}
```

```python
print(f'Hello, {my_name}!')
eligible=True
reasons=[]
if college!='BYU':
  eligible=False
  reasons.append('You must be a BYU graduate')
if years_exp<4:
  eligible=False
  reasons.append('You must have at least 4 years of work experience')
if not pass_fe:
  eligible=False
  reasons.append('You must have passed the fundamentals of engineering (FE) exam')
if eligible:
  print('You are eligible to take the professional engineering (PE) licensing exam.')
else:
  print('You are not eligible to take the professional engineering (PE) licensing exam.')
  print('Reasons:')
  for reason in reasons:
    print(f'- {reason}')
```

3) Input boxes for name and cell number

Then in the same code block, add python code to calculate the price of the pizza. First, start with a base price depending on the size (small = \$12, medium = \$14, large = \$16). Then the first two toppings are free but there is a \$0.50 charge for each topping after that. Finally, add a 5% sales tax.

Print the resulting price using an f print statement. Include the person's name.

Edit the form properties so that the code block is automatically executed whenever a change is made to one of the form inputs.

```python
# @title 🍕 Pizza Order Form
size = "small"  # @param ["small", "medium", "large"] {type:"string"}

# Toppings (each is a checkbox)
cheese=True # @param {type:"boolean"}
pepperoni=False # @param {type:"boolean"}
Ham=True # @param {type:"boolean"}
pineapple=True # @param {type:"boolean"}
mushrooms=False # @param {type:"boolean"}
olives=True # @param {type:"boolean"}
peppers=False # @param {type:"boolean"}
onions=False # @param {type:"boolean"}

name="Kevin Campbell" # @param {type:"string"}
cell="1234567891" # @param {type:"string"}

if size == "small":
  base_price=12
elif size == "medium":
  base_price=14
else:
  base_price=16

toppings=[cheese, pepperoni, Ham, pineapple, mushrooms, olives, peppers, onions]
num_toppings=sum(toppings)

extra_toppings=max(0,num_toppings-2)
topping_cost=0.5*extra_toppings
subtotal=base_price+topping_cost
tax=subtotal*0.05
total=subtotal+tax

print(f"Hi {name}! Your {size} pizza cost ${total:.2f}(including tax).")
```
