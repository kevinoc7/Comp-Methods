## Part 2 - Calculating beam deflections

In this part, you will create and edit a list of beam lengths and then use a nested loop to calculate how much each beam deflects based on the application of a set of loads. Refer to the diagram and equation in the homework description.

```python
# Code Block 5
# Run this cell to reset the same list of beam lengths in cm.
beam_lengths = [830, 420, 900, 360, 640]
```

```python
# Code Block 6
#removes the 5th item from the list which
beam_lengths.pop(4)
print(beam_lengths)
```

```python
# Code Block 7

# Here are the given values for E, I, and the loads given that it's a steel beam.
# Run this cell to initialize the variables and the list

E = 20e6 # Young's modulus in N/cm^2)
I = 4800 # Moment of Inertia in cm^4
loads = [(2500, 150), (7000, 320)] # List of loads (P) in Newtons and their positions (x) in cm on the beams
```

```python
# Code Block 8
#defines the equation to calculate deflection
def calculate_deflection(length, load):
    P, a = load
    delta = (P * a**2 / (6 * E * I)) * (3*length - a)
    return delta

#sets initial deflection to 0
for length in beam_lengths:
  total_deflection=0
#tells how to calculate the deflection for each beam
  for load in loads:
    deflection = calculate_deflection(length,load)
    total_deflection += deflection
#prints results
print('Beam length:', length, 'cm')
print('Total deflection:', total_deflection, 'cm')



```
