```python
####  Create your dictionary here  ####
# Problem 1: Create MaterialsDictionary

MaterialsDictionary = {
    "Guide": ["$ per ton", "$ ship per mile"],
    "Steel": [700, 5],
    "Cement": [130, 2],
    "Lumber": [34, 4],
    "Glass": [50, 15]
}

print(MaterialsDictionary)


#the first word is the material, the first number is the cost of each material per ton
#and the second number is the cost to ship each material per mile
```

```python
####  Start writing your function here  ####

def invoice_calc(MaterialsDictionary, amountmat):
    """Calculate the total cost of a shipment of materials."""
    total_cost = 0

    for i in range(amountmat):
        material = input("Enter material name: ")
        tons = float(input(f"Enter tons of {material}: "))
        miles = float(input(f"Enter miles to ship {material}: "))

        # Retrieve cost per ton and shipping cost per mile
        cost_per_ton = MaterialsDictionary[material][0]
        ship_per_mile = MaterialsDictionary[material][1]

        # Calculate total for this material
        material_cost = (cost_per_ton * tons) + (ship_per_mile * miles)
        total_cost += material_cost

        print(f"Cost for {material}: ${material_cost:.2f}\n")

    print(f"Total shipment cost: ${total_cost:.2f}")
    return total_cost


# Example usage:
# total = invoice_calc(MaterialsDictionary, 2)
# print(f"Total shipment cost: ${total:.2f}")

  # Write the first for loop starting here

def invoice_calc(MaterialsDictionary):
    """Display cost per ton and shipping cost per mile for each material."""

    for key in MaterialsDictionary:
        if key != "Guide":
            cost_per_ton = MaterialsDictionary[key][0]
            ship_per_mile = MaterialsDictionary[key][1]

            print(f"{key}: ${cost_per_ton} per ton, ${ship_per_mile} ship per mile")

  # Write the while loop starting here
def invoice_calc(MaterialsDictionary, amountmat):
    """Calculate the total cost of a shipment of materials using a while loop."""
    total = 0

    while amountmat > 0:
        material = input("Enter material name: ").capitalize()
        quantity = float(input(f"Enter tons of {material}: "))
        distance = float(input(f"Enter miles to ship {material}: "))

        if material not in MaterialsDictionary:
            print("Material not found. Please enter a valid material.\n")
            continue

        cost_per_ton = MaterialsDictionary[material][0]
        ship_per_mile = MaterialsDictionary[material][1]

        item_cost = (cost_per_ton * quantity) + (ship_per_mile * distance)
        total += item_cost

        print(f"Cost for {material}: ${item_cost:.2f}\n")

        amountmat -= 1

    print(f"Total shipment cost: ${total:.2f}")
    return total
```
