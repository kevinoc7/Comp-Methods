## Translator Functions

```python
# 1.2 - Put your text_to_morse function here

MORSE2TEXT_DICT = {v: k for k, v in TEXT2MORSE_DICT.items()}

def text_to_morse(text):
    """Convert normal text to Morse code."""
    text = text.upper()
    words = text.split(" ")
    morse_words = []

    for word in words:
        morse_chars = []
        for char in list(word):
            if char in TEXT2MORSE_DICT:
                morse_chars.append(TEXT2MORSE_DICT[char])
            else:
                morse_chars.append("<ERROR>")
        morse_words.append(" ".join(morse_chars))
    return "   ".join(morse_words)





```

```python
# 1.3 - Put your morse_to_text function here
def morse_to_text(morse):
    """Convert Morse code back to normal text."""
    words = morse.split("   ")
    text_words = []

    for word in words:
        chars = word.split(" ")
        text_chars = []
        for char in chars:
            if char in MORSE2TEXT_DICT:
                text_chars.append(MORSE2TEXT_DICT[char])
            else:
                text_chars.append("<ERROR>")
        text_words.append("".join(text_chars))
    return " ".join(text_words)
```

# PART 2 - Rebar Cut Optimizer

## Function for sorting rebar cuts

```python
# 2.1 - Write your sort_cuts function here
def sort_cuts(cut_lengths, stock_length):
    """
    Create a cut plan for rebar stock using a greedy algorithm.
    Each sublist in cut_plan represents cuts made from one stock length.
    Returns: cut_plan (list of lists), waste (float), efficiency (float)
    """
    # Sort cuts from longest to shortest (greedy approach)
    cuts = sorted(cut_lengths, reverse=True)
    cut_plan = []  # list of lists, each representing one stock piece

    # Iterate through each cut length
    for cut in cuts:
        placed = False
        # Try to fit the cut into an existing stock piece
        for plan in cut_plan:
            if sum(plan) + cut <= stock_length:
                plan.append(cut)
                placed = True
                break
        # If no existing stock piece can fit the cut, start a new one
        if not placed:
            cut_plan.append([cut])

    # Calculate total waste and efficiency
    total_stock = len(cut_plan) * stock_length
    total_cuts = sum(sum(plan) for plan in cut_plan)
    waste = total_stock - total_cuts
    efficiency = (total_cuts / total_stock) * 100

    return cut_plan, waste, efficiency

```

## Function for optimizing cuts

```python
# 2.4 - Write your optimize_rebar function here
def optimize_rebar(cut_lengths, min_stock, max_stock, increment):
    """
    Tests multiple stock lengths to find the one with minimum waste and maximum efficiency.
    Returns: best_cut_plan, best_stock_length, best_waste, best_efficiency
    """
    best_efficiency = 0
    best_waste = float('inf')
    best_cut_plan = []
    best_stock_length = None

    # Loop through stock lengths from min_stock to max_stock
    for stock_length in range(min_stock, max_stock + 1, increment):
        cut_plan, waste, efficiency = sort_cuts(cut_lengths, stock_length)
        print(f"{stock_length:.2f} ft -> waste={waste:.2f} ft, efficiency={efficiency:.2f}%")

        # Update best plan if efficiency improves or waste decreases
        if efficiency > best_efficiency or (efficiency == best_efficiency and waste < best_waste):
            best_efficiency = efficiency
            best_waste = waste
            best_cut_plan = cut_plan
            best_stock_length = stock_length

    # Print optimal results
    print(f"\nBest stock length = {best_stock_length:.1f} ft")
    print(f"Number of stock required = {len(best_cut_plan)}")
    return best_cut_plan, best_stock_length, best_waste, best_efficiency

```
