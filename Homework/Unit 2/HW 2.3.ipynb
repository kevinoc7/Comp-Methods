# Part 1 - Truss Determinacy Solver

```python
joints = int(input("Enter the number of joints in the truss: "))
members = int(input("Enter the number of members in the truss: "))
reactions = int(input("Enter the number of reaction forces acting on the truss: "))

if members+reactions==2*joints:
  result="statically determinant"
elif members+reactions>2*joints:
  result="statically indeterminant"
else:
  result="unstable"

print(result)

print(f"The truss is {result}.")






```

# Part 2 - Material Order Counter

```python
# List of orders
orders = [
    ['wood', 'steel', 'glass', 'steel'],
    ['brick', 'wood', 'steel', 'wood'],
    ['glass', 'steel', 'wood', 'concrete'],
    ['wood', 'brick', 'wood', 'brick'],
    ['concrete', 'concrete', 'glass'],
    ['wood', 'concrete', 'glass'],
    ['concrete', 'glass', 'wood', 'steel'],
    ['brick', 'glass', 'brick', 'steel'],
    ['concrete', 'brick', 'brick', 'wood'],
    ['brick', 'glass', 'wood', 'wood', 'steel'],
    ['glass', 'glass', 'brick', 'brick', 'glass'],
    ['wood', 'wood', 'wood', 'steel', 'glass', 'glass']
]

# Materials and starting totals
Total_order = [['wood', 0], ['brick', 0], ['steel', 0], ['glass', 0], ['concrete', 0]]
#basically for each item in the list it adds one to the count of that specific material
for order in orders:
    for item in order:
        for material in Total_order:
            if item == material[0]:
                material[1] += 1
#prints the counted totals of each material in the list comprehensively
for material in Total_order:
    print(f"{material[1]} orders of {material[0]}")
```
