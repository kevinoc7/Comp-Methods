### Part 1a - Shear Diagram

You will create your shear diagram in the code block below.


```python

L = 6      # ft
RA = 170   # lb
RB = 410   # lb
P1 = 100   # lb
q1 = 120   # lb/ft
q2 = 120   # lb/ft

x = np.linspace(0, L, 500)

V = np.zeros_like(x)

for i in range(len(x)):
    if x[i] < 2:
        V[i] = RA - q1 * x[i]

    elif x[i] < 4:
        V[i] = RA - (q1 * 2) - P1

    else:
        V[i] = RA - (q1 * 2) - P1 + RB - q2 * (x[i] - 4)

plt.figure(figsize=(10,6))
plt.plot(x, V, linewidth=2)

plt.fill_between(x, V, 0, alpha=0.3)

plt.axhline(0, color='black')

plt.title('Shear Diagram')
plt.xlabel('x (ft)')
plt.ylabel('V (lb)')
plt.grid(True)

plt.show()
```

### Part 1b - Moment Diagram

You will create your moment diagram in the code block below.

```python

L = 6      # ft
RA = 170   # lb
RB = 410   # lb
P1 = 100   # lb
q1 = 120   # lb/ft
q2 = 120   # lb/ft

x = np.linspace(0, L, 500)

M = np.zeros_like(x)

for i in range(len(x)):
    if x[i] < 2:
        M[i] = (RA * x[i]) - (q1 * x[i]**2 / 2)

    elif x[i] < 4:
        M[i] = (RA * x[i]) \
               - (q1 * 2 * (x[i] - 1)) \
               - (P1 * (x[i] - 2))

    else:
        M[i] = (RA * x[i]) \
               - (q1 * 2 * (x[i] - 1)) \
               - (P1 * (x[i] - 2)) \
               + (RB * (x[i] - 4)) \
               - (q2 * (x[i] - 4) * ((x[i] - 4) / 2))

plt.figure(figsize=(10,6))
plt.plot(x, M, linewidth=2, color='red')

plt.fill_between(x, M, 0, color='red', alpha=0.3)

plt.axhline(0, color='black')

plt.title('Moment Diagram')
plt.xlabel('x (ft)')
plt.ylabel('M (lb-ft)')
plt.grid(True)

plt.show()
```

### Part 5 - Average U.S. Daily Birth Data

You will now create a graph using average daily birth data in the U.S.

### Numpy Array Creation & Slicing

Slice the array to create the x and y-axis values. These will correspond to the following columns in the CSV linked on the HW instructions page:

*   x - Date
*   y_1 - Births

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = np.genfromtxt(
    '/content/2012-US-Birth-Rates.csv',
    delimiter=',',
    skip_header=1,
    dtype=str
)

x = pd.to_datetime(data[:, 0])
y_1 = data[:, 1].astype(float)
```

### Graphing the Data

Plot x and y_1 with the elements shown in the instructions in the code block below.

```python
plt.figure(figsize=(12, 4))

plt.plot(x, y_1, color='turquoise', linewidth=1.5)

plt.title('Average U.S. Daily Birth Data')
plt.xlabel('Date')
plt.ylabel('Births')

plt.grid(True)

plt.annotate(
    "New Year's Day",
    xy=(pd.Timestamp('2012-01-01'), 4000),
    xytext=(pd.Timestamp('2012-01-15'), 3900),
    arrowprops=dict(arrowstyle='->',
                    connectionstyle='arc3,rad=-0.3',
                    lw=2)
)

plt.annotate(
    '4th of July',
    xy=(pd.Timestamp('2012-07-04'), 4300),
    xytext=(pd.Timestamp('2012-08-05'), 4200),
    arrowprops=dict(arrowstyle='->',
                    connectionstyle='arc3,rad=-0.4',
                    lw=2)
)

plt.annotate(
    'Halloween',
    xy=(pd.Timestamp('2012-10-31'), 4680),
    xytext=(pd.Timestamp('2012-09-15'), 4600),
    arrowprops=dict(arrowstyle='->',
                    connectionstyle='arc3,rad=-0.4',
                    lw=2)
)

plt.annotate(
    'Thanksgiving',
    xy=(pd.Timestamp('2012-11-22'), 4475),
    xytext=(pd.Timestamp('2012-10-12'), 4280),
    arrowprops=dict(arrowstyle='->',
                    connectionstyle='arc3,rad=0.4',
                    lw=2)
)

plt.annotate(
    'Christmas',
    xy=(pd.Timestamp('2012-12-25'), 3850),
    xytext=(pd.Timestamp('2012-10-22'), 3900),
    arrowprops=dict(arrowstyle='->',
                    connectionstyle='arc3,rad=0.5',
                    lw=2)
)

plt.tight_layout()
plt.show()

# Birth rates may drop on major holidays because fewer scheduled
# births and medical procedures are performed on those days.
# Hospitals often postpone elective inductions and C-sections until
# after the holiday, reducing the number of births recorded.
```
