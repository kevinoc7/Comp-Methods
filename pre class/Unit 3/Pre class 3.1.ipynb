**Code Block 2**


1.   Turn list A into a numpy array named Array_A
2. Turn list B into a numpy array named Array_B
3. Create a new array named Array_C by appending Array_A to Array_B.


---



```python
List_A = [1,2,3,4,5]
List_B = [6,7,8,9,10]
Array_A = np.array(List_A)
Array_B = np.array(List_B)
Array_C = np.append(Array_A, Array_B)
print(Array_C)
```

**Code Block 3**



1. Create a 2-D array from lists A_1 and A_2. Name it Array_A.
2. Make a copy of Array_A and name it Array_B. HINT: Use the np.copy() function.
3. Use np.linalg.solve() with Array_A and Array_B to solve the system of linear equations.
4. Print the solution.




---



```python
A_1 = [1,2]
A_2 = [3,4]
Array_A = np.array([A_1, A_2])
Array_B = np.copy(Array_A)
solution = np.linalg.solve(Array_A, Array_B)
print(solution)
```
