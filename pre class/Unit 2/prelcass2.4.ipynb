```python
rivers={}

while True:
  river=input("Enter the name of the river:")
  country=input(f"Enter the name of the country where the {river} runs:")

  rivers[river]=country
  add_more=input("Do you want to add another river? (yes/no):")
  if add_more =="no":
      break

print("\nRivers and their countries:")
for river, country in rivers.items():
    print(f"The {river} runs through {country}.")

print("\nRivers:")
for river in rivers.keys():
    print(river)

print("\nCountries:")
for country in rivers.values():
    print(country)
```
