```python

def make_album(artist, title, num_songs=None):
    """Build a dictionary describing a music album."""
    album = {
        'artist': artist.title(),
        'title': title.title()
    }
    if num_songs is not None:
        album['num_songs'] = num_songs
    return album

album1 = make_album('taylor swift', 'midnights')
album2 = make_album('john mayer', 'continuum')
album3 = make_album('coldplay', 'parachutes', num_songs=10)

print(album1)
print(album2)
print(album3)
```
